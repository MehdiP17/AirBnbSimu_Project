```python
import logging
import boto3
import os
import io
import pandas as pd
from dotenv import load_dotenv

logging.getLogger('botocore').setLevel(logging.INFO)
logging.getLogger('boto3').setLevel(logging.INFO)

load_dotenv(override=True)

ACCESS_KEY_ID = os.getenv('AWS_ACCESS_KEY_ID')
SECRET_ACCESS_KEY = os.getenv('AWS_SECRET_ACCESS_KEY')
BUCKET_NAME = os.getenv('AWS_BUCKET_NAME')
REGION_NAME = os.getenv('AWS_REGION_NAME')
FOLDER = "inside-airbnb/"

session = boto3.Session(
    region_name=REGION_NAME,
    aws_access_key_id=ACCESS_KEY_ID, 
    aws_secret_access_key=SECRET_ACCESS_KEY
)

s3 = session.client('s3')
response = s3.list_objects_v2(Bucket=BUCKET_NAME, Prefix = FOLDER)
files = response.get('Contents', [])


listings = [f for f in files if 'listings' in f['Key'] and f['Key'].endswith('.csv')]

dataframes = {}

for f in listings:
    obj = s3.get_object(Bucket=BUCKET_NAME, Key=f['Key'])
    dataframes[f['Key']] = pd.read_csv(io.BytesIO(obj['Body'].read()))
    print("Fichier chargé :", f['Key'])
```
